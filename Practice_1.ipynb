```python
# Data Types
  # 1) Primtive Data Types - Which Stores Single Value.  Ex- int,float,String,Boolean etc.
  # 2) Non Primitive Data Types - Which Stores Collection Of Values.  Ex - List, Dic, Tuple, Set etc.
```

```python
# Object type
#  1) Mutable - Can Be Change.  Ex- list,dic,set [LDS]
#  2) immutable - cant Be Change.  Ex - Strigs, Int, tuple [SIT]

# 1. List - Mutable
fruits = ['Apple','Mango','Banana']
print(fruits[0]) # Access the Value
fruits[1]='Orange'# Change the Value
print(fruits)
fruits.append('Grapes') # Add the Value
print(fruits)
fruits.remove('Apple') # Remove the Value
print(fruits)

# 2. Dictionary - Mutable
person ={
    'name': 'Aniket',
    'age': 25,
    'City': 'Mumbai'
}
print(person['name']) # Access Dic
person['Salary']=30000 # Change Value
print(person)

# 3 . Set - Mutable
# No Duplicates that Removes Automatic
# No Ordered # No Indexing
num = {10,20,30,40}
print(num)
num.add(50) # Add Value
print(num)
num.remove(10) # Remove Value
print(num)

# 4) Tuple - Immutable
# Ordered
# Duplicates Allow
# Stores Diff Data Types
# EX 1. -
# data = ('Aniket','30','2.03',True')
# EX 2 -
Fruits =('apple','Banana','Mango')
print(Fruits[0])    # Indexing
print (Fruits[0:2]) # Slicing
```

```python
# 4) Membership Operators
# 1. in - Value Exist ?
# 2. not in - Value not Exist ?

# Ex 1
fruits = ('Apple','Mango','Banana')
print('Apple' in fruits)
print('Orange'not in fruits)

#Ex 2
num = [20,103,150]
print(20 in num)
print(500 not in num)

```
